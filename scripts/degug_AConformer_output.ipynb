#### Check that every msa file have a sequence name 

```julia
function check_msa_files(filenames)
    list=[]
    println("🔍 Vérification du fichier : $filenames")
    open(filenames, "r") do f
        lines = readlines(f)
        has_error = false
        for i in 1:length(lines)
            if startswith(lines[i], ">") 
                push!(list,lines[i])
                if length(lines[i])<2
                    println(lines[i])
                    println("❌ Erreur : Une séquence sans en-tête trouvée à la ligne $i dans $filenames")
                    has_error = true
                end
                
            end
        end
        if !has_error
            println("✅ Fichier MSA correct : toutes les séquences ont un nom.")
        end
    end
    return list
    
end
```

```julia
path ="/store/EQUIPES/AMIG/MEMBERS/julie.daniel/AlphaConformers.jl/data"
l_msas=[]
for f in readdir(path; join=true)  # Liste les fichiers avec le chemin complet
    if isdir(f)
        path_file=f*"/afdb_up_results/msa.a3m"
        println(path_file)
        list_msa=check_msa_files(path_file)
        println(list_msa)
        println(length(list_msa))
        push!(l_msas,list_msa)
        path_file=f*"/afdb_up_results/msa.a3m.bak"
        println(path_file)
        list_bak=check_msa_files(path_file)
        println(list_bak)
        println(length(list_bak))
    end
    
end
println(l_msas)
```

#### Compare the msa that work with the other 

```julia
using DeepDiffs

function parse_msa(file)
    msa_dict = Dict()
    open(file, "r") do f
        current_name = ""
        for line in eachline(f)
            if startswith(line, ">")
                current_name = strip(line)  # Nettoie le nom
                msa_dict[current_name] = ""
            else
                msa_dict[current_name] *= strip(line)  # Ajoute la séquence au bon en-tête
            end
        end
    end
    return msa_dict
end

function compare_msa(file1, file2)
    println("🔍 Comparaison de $file1 et $file2")
    msa1 = parse_msa(file1)
    msa2 = parse_msa(file2)
    diff_result = deepdiff(msa1, msa2)

    if isempty(diff_result)
        println("✅ Les fichiers sont identiques !")
    else
        println("⚠️ Différences détectées entre $file1 et $file2 :")
        println(diff_result)
    end
end

# Comparer les fichiers entre eux
compare_msa("msa1.a3m", "msa2.a3m")
compare_msa("msa1.a3m", "msa3.a3m")
compare_msa("msa1.a3m", "msa4.a3m")
compare_msa("msa1.a3m", "msa5.a3m")
```
